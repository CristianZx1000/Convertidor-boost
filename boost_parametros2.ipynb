#### PI

```python
f_s = 100e3 # Hz / PWM

Vi = 5 # V
Vo = 20 # V
Po = 50 # W
R = Vo**2/Po # Ohm
print(f"R = {R:.3f}")

D = 1 - Vi/Vo

Lmin = D * (1 - D)**2 * R/(2 * f_s)
print(f"Lmin = {Lmin*1e6:.2f} (\u03bcH)")
delta_V = Vo * 1/100 # V 
Cmin = D/(R * f_s * delta_V)
print(f"Cmin = {Cmin*1e6:.2f} (\u03bcF)")

############################################
RL = 1e-3 # Ohm
L = 120e-6 # H
C = 1 * 100e-6 # F
R_mosfet = 0.85 #1e-3 # Ohm

R_tot = RL + R_mosfet # LVK: RL - mosfet (estado:on)

# Lazo control de corriente
I_kp = L * f_s/(2 * Vo)  
I_ki = R_tot * f_s/(2 * Vo)

print(f"I_kp = {I_kp:.6f}")
print(f"I_ki = {I_ki:.6f}")

# Lazo control de voltaje
V_kp = C * f_s/2 #/250
V_ki = (C * f_s**2)/8 #/250*250

print(f"V_kp = {V_kp:.6f}")
print(f"V_ki = {V_ki:.6f}")
```
