```python
import math

f_s = 100e3      # Hz / PWM
Vi = 5.0         # V (nominal)
Vo = 20.0        # V (referencia)
Po = 50.0        # W
R = Vo**2 / Po   # 8.0 Ohm
D = 1.0 - (Vi / Vo)

RL = 1e-3        # Ohm
R_mosfet = 5e-3  # Ohm (R_on para 50 W)
R_tot = RL + R_mosfet
L = 120e-6       # H
C = 100e-6   # F

# 1. Verificación del cero en el semiplano derecho (RHPZ)
f_rhpz = (R * (1.0 - D)**2) / (2.0 * math.pi * L)
print(f"Frecuencia límite RHPZ = {f_rhpz:.1f} Hz")

# 2. Lazo interno de corriente (f_ci = f_s / 15 = 6.67 kHz)
f_ci = f_s / 15.0
w_ci = 2.0 * math.pi * f_ci
I_kp = (w_ci * L) / Vo
I_ki = I_kp * (w_ci / 5.0)

print(f"I_kp = {I_kp:.6f} | I_ki = {I_ki:.6f}")

# 3. Lazo externo de voltaje (f_cv <= f_rhpz / 4, ej. 100 Hz)
f_cv = min(100.0, f_rhpz / 4.0)
w_cv = 2.0 * math.pi * f_cv
V_kp = (w_cv * C) / (1.0 - D)
V_ki = V_kp * (w_cv / 4.0)

print(f"V_kp = {V_kp:.6f} | V_ki = {V_ki:.6f}")
```
